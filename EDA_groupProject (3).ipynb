```python
## Distribution of Average Price

plt.figure()
plt.hist(df['Average_Price_USD'])
plt.xlabel("Average Price (USD)")
plt.ylabel("Frequency")
plt.title("Average Price Distribution")
plt.show()
```

## BIVARIATE ANALYSIS

```python
## Carbon Footprint vs Water Usage

plt.figure()
plt.scatter(df['Carbon_Footprint_MT'], df['Water_Usage_Liters'])
plt.xlabel("Carbon Footprint (MT)")
plt.ylabel("Water Usage (Liters)")
plt.title("Carbon Footprint vs Water Usage")
plt.show()
```

```python
## Sustainability Rating vs Average Price

plt.figure()
sns.boxplot(x='Sustainability_Rating', y='Average_Price_USD', data=df)
plt.title("Sustainability Rating vs Average Price")
plt.show()
```

```python
## Market Trend vs Recycling Programs

plt.figure()
sns.countplot(x='Market_Trend', hue='Recycling_Programs', data=df)
plt.title("Market Trend vs Recycling Programs")
plt.show()
```

## MULTIVARIATE ANALYSIS

```python
## Correlation Heatmap

num_cols = ['Carbon_Footprint_MT','Water_Usage_Liters','Waste_Production_KG','Product_Lines','Average_Price_USD']
plt.figure(figsize=(8,6))
sns.heatmap(df[num_cols].corr(), annot=True)
plt.title("Correlation Heatmap")
plt.show()
```

```python
## Pair Plot

sns.pairplot(df[num_cols])
plt.show()
```
